CSPP repersentation

```python
COLORS = ['C','M','Y','K']

def make_domains(variables, colors):
  return {var: set(colors) for var in variables}

def constraint(xi, a, xj, b):
  return a!=b
```

```python
def revise(domains, xi, xj):
  revised = False
  for a in set(domains[xi]):
    if not any(constraint(xi,a,xj,b) for b in domains[xj]):
      domains[xi].remove(a)
      revised = True
  return revised
```

AC3

```python
def ac3(domains, neighbours, queue = None):
  if queue is None:
    queue = deque((xi,xj) for xi in domains for xj in neighbours[xi])
  while queue:
    xi, xj = queue.popleft()
    if revise(domains, xi, xj):
      if len(domains[xi]) == 0:
        return False
      for xk in neighbours[xi]:
        if xk != xj:
          queue.append((xk,xi))
  return True
```

```python
domains = make_domains(district_names, COLORS)
result = ac3(domains, district_neighbours)
result
```
